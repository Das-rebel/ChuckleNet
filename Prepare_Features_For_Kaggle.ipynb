# Prepare features_255 for Kaggle Upload

**What this does:**
1. Mounts Google Drive
2. Verifies the features_255 folder exists
3. Zips it for download
4. You then upload the zip to Kaggle

```
# Cell 1: Mount Drive + Verify Features
from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import os

FEATURE_DIR = '/content/drive/MyDrive/standup4ai/features_255'
if not os.path.exists(FEATURE_DIR):
    raise FileNotFoundError(f'Not found: {FEATURE_DIR}')

feat_files = [f for f in os.listdir(FEATURE_DIR) if '_features.npy' in f]
lbl_files = [f for f in os.listdir(FEATURE_DIR) if '_labels.npy' in f]
print(f'Features: {len(feat_files)}')
print(f'Labels:   {len(lbl_files)}')

if len(feat_files) != 49:
    print(f'WARNING: Expected 49, got {len(feat_files)}')
    print('Either Batch 1 incomplete or extra files.')
```

```
# Cell 2: Zip the features
import shutil

# Zip with directory structure preserved
shutil.make_archive('/content/features_255', 'zip', root_dir='/content/drive/MyDrive/standup4ai',
                    base_dir='features_255')

# Verify
zip_size = os.path.getsize('/content/features_255.zip')
print(f'Created: /content/features_255.zip ({zip_size/1e6:.1f} MB)')
```

```
# Cell 3: Download zip to local machine
from google.colab import files
files.download('/content/features_255.zip')

print('Download started. Check your browser downloads folder.')
print('Next step: Upload features_255.zip to Kaggle as a dataset.')
```
